
````python
from ultralytics import YOLO

# Sử dụng mô hình YOLOv8n (Bản nano nhẹ nhất, phù hợp cho real-time giao thông)
model = YOLO("yolov8n.pt")

# Tiến hành Train
results = model.train(
    data="/content/vvd_yv8_dataset/data.yaml", # Đường dẫn tuyệt đối đến file yaml vừa sửa
    epochs=50,                                 # Chạy thử nghiệm 50 lượt học
    imgsz=640,                                 # Kích thước ảnh chuẩn của YOLOv8
    plots=True                                 # Vẽ biểu đồ theo dõi độ chính xác
)
````
